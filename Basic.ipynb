```python
# Using a for loop from 1 to 50:

# If a number is divisible by both 3 and 5, print "FizzBuzz".
# If divisible by only 3, print "Fizz".
# If divisible by only 5, print "Buzz".
# Otherwise print the number.


for i in range(1,10):
  if i%3==0 and i%5==0:
    print("Fizzbuzz");
  elif i%3==0:
    print("Fizz");
  elif(i%5==0):
    print("Buzz");
  else:
    print(i)
```

```python
# If B = "asdfghjkl" and the keys a, s, l, and g on the keyboard are damaged (not working), what will be printed when you try to type the string B?

B="asdfghjkl"

for alpha in B:
  if alpha=="a":
    continue
print(B)
```

```python

users = [
    {"id": 1, "name": "Joy"},
    {"id": 2, "name": "Rahim"},
    {"id": 3, "name": "Karim"},
    {"id": 4, "name": "Nayeem"}
]


print("*******Register Please*********")
reg_name=input("Enter User name:")
reg_pass=input("Password:")

print("*****login*****")
log_name=input("Enter user name:")
log_pass=input("Password:")

if reg_name==log_name and reg_pass==log_pass:
    print("Coose Option")
    print(
    "1. All Users\n"
    "2. Find User By ID\n"
    "3. Add Users\n"
    "4. Remove Users\n"
    "5. Show Users Name\n"
    "6.Find Users By Name\n\n"
)
menu=input("********Choose One*******\n\n")
if menu=="1":
    print("#All Users")
    print("ID  Name")
    for Users in users:
      print(Users['id']," ",Users["name"])


elif menu=="2":
    print("#Find User By ID")
    fin_id=int(input("Enter ID: "))
    result=list(filter(lambda u :u["id"]==fin_id,users))
    if result:
        print("ID:", result[0]["id"])
        print("Name:", result[0]["name"])
    else:
      print("User Not Exist!")


elif menu=="6":
  print("Find Users By Name")
  fin_name=input("Enter Name: ")
  result=list(filter(lambda n: n["name"]==fin_name,users))
  if result:
    print("ID:", result[0]["id"])
    print("Name:", result[0]["name"])
  else:
    print("User Not Exist!")



else:
  print("Login pass or user name not correct")
```

```python
# Using a for loop from 1 to 50:

# If a number is divisible by both 3 and 5, print "FizzBuzz".
# If divisible by only 3, print "Fizz".
# If divisible by only 5, print "Buzz".
# Otherwise print the number.


for i in range(1,10):
  if i%3==0 and i%5==0:
    print("Fizzbuzz");
  elif i%3==0:
    print("Fizz");
  elif(i%5==0):
    print("Buzz");
  else:
    print(i)
```

```python
users = [
    {"id": 1, "name": "Joy"},
    {"id": 2, "name": "Rahim"},
    {"id": 3, "name": "Karim"},
    {"id": 4, "name": "Nayeem"}
]

print("*******Register Please*********")
reg_name = input("Enter User name: ")
reg_pass = input("Password: ")

print("*****Login*****")
log_name = input("Enter user name: ")
log_pass = input("Password: ")

if reg_name == log_name and reg_pass == log_pass:

    while True:
        print("\n===== MENU =====")
        print("1. All Users")
        print("2. Find User By ID")
        print("3. Add User")
        print("4. Remove User")
        print("5. Show User Names")
        print("6. Find User By Name")
        print("7. Exit")

        menu = input("Choose One: ")

        if menu == "1":
            print("\nID  Name")
            for user in users:
                print(user["id"], user["name"])

        elif menu == "2":
            fin_id = int(input("Enter ID: "))
            result = [u for u in users if u["id"] == fin_id]

            if result:
                print(result[0])
            else:
                print("User Not Exist!")

        elif menu == "3":
            new_user_id = int(input("Enter ID: "))
            new_user_name = input("Enter Name: ")

            users.append({
                "id": new_user_id,
                "name": new_user_name
            })

            print("User Added Successfully")

        elif menu == "4":
            rem_id = int(input("Enter ID: "))
            result = [u for u in users if u["id"] == rem_id]

            if result:
                users.remove(result[0])
                print("User Removed Successfully")
            else:
                print("User Not Exist!")

        elif menu == "5":
            print("\nUser Names:")
            for user in users:
                print(user["name"])

        elif menu == "6":
            fin_name = input("Enter Name: ")
            result = [u for u in users if u["name"].lower() == fin_name.lower()]

            if result:
                print(result[0])
            else:
                print("User Not Exist!")

        elif menu == "7":
            print("Goodbye!")
            break

        else:
            print("Invalid Option!")

else:
    print("Login username or password not correct!")
```
